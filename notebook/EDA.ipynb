```python
###  take  the  missing    columns only   
miss_cols   =   [cols  for cols  in  data.columns   if  data[cols].isnull().sum()  > 1 ]
```

```python
miss_cols
```

```python
###  take  he  missing   values  object  columns    

miss_cols_obj   =  [cols   for cols  in   miss_cols  if  data[cols].dtypes  == "O"]
```

```python
miss_cols_obj
```

```python
###  take  the  value  conts   
for cols  in  miss_cols_obj  :  
    print(data[cols].value_counts())   
    print()  ##  this  is  the  new  line 
```
