```python


# Check your categorical columns with missing values
have_pattern_miss_cols  = ['Safety of care national comparison',
                 'Effectiveness of care national comparison', 
                 'Timeliness of care national comparison']

for col in have_pattern_miss_cols :
    print(f"{col}: {data[col].unique()}")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder

# numeric features
numeric_features = data.select_dtypes(include=['number']).columns.tolist()
numeric_features = [f for f in numeric_features if f != 'Hospital overall rating']

print("Features used:", numeric_features)
```

```python
for col in have_pattern_miss_cols:
    print(f"\n{'='*50}")
    print(f"Imputing: {col}")
    
    # Split known and unknown
    known_data   = data[data[col].notna()]
    unknown_data = data[data[col].isna()]
    
    print(f"  Known rows   : {len(known_data)}")
    print(f"  Missing rows : {len(unknown_data)}")
    print(f"  Categories   : {known_data[col].unique()}")
    
    # Encode string → number
    le = LabelEncoder()
    y_train = le.fit_transform(known_data[col])
    
    # Check what LabelEncoder assigned
    print(f"  Encoding     : {dict(zip(le.classes_, le.transform(le.classes_)))}")
    
    # Prepare features
    X_train = known_data[numeric_features].fillna(0)
    X_pred  = unknown_data[numeric_features].fillna(0)
    
    # Train
    clf = RandomForestClassifier(n_estimators=100, random_state=42)
    clf.fit(X_train, y_train)
    
    # Predict
    predicted        = clf.predict(X_pred)
    predicted_labels = le.inverse_transform(predicted)
    
    # Fill
    data.loc[data[col].isna(), col] = predicted_labels
    
    print(f"  Filled Sample: {predicted_labels[:5]}")

print("\n✅ Done!")
```

```python
import scipy.stats as stats

numerical_miss_cols = ['HCAHPS Linear Mean Value', 
                       'Number of Completed Surveys', 
                       'Survey Response Rate Percent']

for col in numerical_miss_cols:
    
    # Create missingness flag (same as before)
    data_copy = data.copy()
    data_copy[col + '_missing'] = np.where(data[col].isnull(), 1, 0)
    
    # Split into two groups
    missing_group    = data_copy[data_copy[col + '_missing'] == 1]['Hospital overall rating']
    not_missing_group= data_copy[data_copy[col + '_missing'] == 0]['Hospital overall rating']
    
    print(f"\n{'='*50}")
    print(f"Column: {col}")
    print(f"Mean rating (Present) : {not_missing_group.mean():.3f}")
    print(f"Mean rating (Missing) : {missing_group.mean():.3f}")
    print(f"Difference            : {abs(not_missing_group.mean() - missing_group.mean()):.3f}")
    
    # T-test to check if difference is statistically significant
    t_stat, p_value = stats.ttest_ind(missing_group.dropna(), 
                                       not_missing_group.dropna())
    print(f"P-value               : {p_value:.4f}")
    
    if p_value < 0.05:
        print(f"Result  → ⚠️  MAR/MNAR  (missingness has pattern)")
    else:
        print(f"Result  → ✅  MCAR      (missingness is random)")
    
    # Bar chart (same as before)
    data_copy.groupby(col + '_missing')['Hospital overall rating'].mean().plot.bar(
        color=['green', 'red']
    )
    plt.title(f"{col} vs Hospital overall rating missingness")
    plt.xticks([0, 1], ['Present', 'Missing'], rotation=0)
    plt.show()
```

###  Aggregate  the  column  for  the  Measure ID  

```python
# Pivot: each row becomes a hospital, each measure ID becomes a column with its linear score
df_pivot = df_cleaned.pivot_table(
    index='Facility ID',
    columns='HCAHPS Measure ID',
    values='HCAHPS Linear Mean Value',
    aggfunc='first'          # each hospital has exactly one value per measure
).reset_index()

# Rename columns for clarity (adjust order based on your actual column names)
# The actual names are: 'H_COMP_1_LINEAR_SCORE', 'H_COMP_2_LINEAR_SCORE', 
# 'H_COMP_3_LINEAR_SCORE', 'H_CLEAN_LINEAR_SCORE', 'H_QUIET_LINEAR_SCORE'
df_pivot.rename(columns={
    'H_COMP_1_LINEAR_SCORE': 'nurse_comm',
    'H_COMP_2_LINEAR_SCORE': 'doctor_comm',
    'H_COMP_3_LINEAR_SCORE': 'staff_responsiveness',
    'H_CLEAN_LINEAR_SCORE': 'cleanliness',
    'H_QUIET_LINEAR_SCORE': 'quietness'
}, inplace=True)

# Create composite welcoming_score (average of the five)
df_pivot['welcoming_score'] = df_pivot[['nurse_comm', 'doctor_comm', 
                                        'staff_responsiveness', 'cleanliness', 
                                        'quietness']].mean(axis=1)

print(f"Pivoted shape: {df_pivot.shape}")  # Should be (~3,576 rows, 7 columns)
```

```python
# Group by Facility ID and take the first occurrence of each constant column
constant_cols = [
    'State', 'Hospital Type', 'Hospital Ownership', 'Emergency Services',
    'Hospital overall rating',
    'Safety of care national comparison', 'Patient experience national comparison',
    'Effectiveness of care national comparison', 'Timeliness of care national comparison',
    'Efficient use of medical imaging national comparison'
]

df_constants = df_cleaned.groupby('Facility ID')[constant_cols].first().reset_index()
```

```python
df_metadata = df_cleaned.groupby('Facility ID')[['Number of Completed Surveys', 
                                                  'Survey Response Rate Percent']].mean().reset_index()
```

```python
# Start with the pivoted welcoming scores
df_hospital = df_pivot.copy()

# Merge constants
df_hospital = df_hospital.merge(df_constants, on='Facility ID', how='left')

# Merge metadata
df_hospital = df_hospital.merge(df_metadata, on='Facility ID', how='left')

print(f"Final hospital-level shape: {df_hospital.shape}")
df_hospital.head()
```
